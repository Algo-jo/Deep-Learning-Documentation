```python
import os 
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import keras
from sklearn.model_selection import train_test_split
```

```python
img_row, img_col = 100, 100
x, y = [], []
DATA_PATH = './dataset/'
counter = 0
classes = os.listdir(DATA_PATH)
print(classes)
```

```python
for class_path in classes:
    for img_path in os.listdir(f"{DATA_PATH}{class_path}"):
        path = f"{DATA_PATH}{class_path}/{img_path}"
        label = counter
        img = tf.keras.preprocessing.image.load_img(path, target_size=(img_row, img_col))
        img = tf.keras.preprocessing.image.img_to_array(img)
        x.append(img)
        y.append(label)
    counter += 1

x = np.array(x)
y = np.array(y)
x = x.astype('float32') / 255.0
```

```python
x_train, y_train, x_temp, y_temp = train_test_split(
    x, y, test_size=0.2, random_state=42
)

x_test, y_test, x_val, y_val = train_test_split(
    x_temp, y_temp, test_size=0.5, random_state=42
)

y_train = keras.utils.to_categorical(y_train)
y_test = keras.utils.to_categorical(y_test)
y_val = keras.utils.to_categorical(y_val)
print(x_train.shape)
```

```python
import keras.layers

model = keras.Sequential()

model.add(keras.layers.Conv2D(6, (6,6), activation="relu", input_size=(img_row, img_col, 3)))
model.add(keras.layers.AveragePooling1D((2,2), 2))
model.add(keras.layers.Conv2D(16, (6,6), activation="relu"))
model.add(keras.layers.AveragePooling1D((2,2), 2))

model.add(keras.layers.Flatten())

model.add(keras.layers.Dense(120, activation="relu"))
model.add(keras.layers.Dense(84, activation="relu"))
model.add(keras.layers.Dense(3, activation="softmax"))

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=['accuracy']
)
history = model.fit(
    x_train, y_train,
    batch_size=120,
    epochs=10,
    validation_data=(x_val, y_val)
)
```

```python
model.evaluate(x_test, y_test)
```
