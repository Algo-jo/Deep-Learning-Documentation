```python
import torch
import torch.nn as nn
```

# Multi Layer Perceptron Classifier

```python
class MLPClassifier(nn.Module):
    def __init__(self):
        super().__init__()

        self.net = nn.Sequential(
            nn.Linear(5, 64),  # 5 Input Layer -> 64 Hidden Layer
            nn.ReLU(), # Activation ReLU
            nn.Linear(64, 1), # 64 Hidden Layer -> 1 Output Layer 
            nn.Sigmoid() # Output activation sigmoid for Binary Classification
        )

    def forward(self, x):
        return self.net(x)
```

```python
model = MLPClassifier()
loss_fn = nn.BCELoss() # Loss = Binary Cross Entropy 
optimizer = torch.optim.Adam(model.parameters(), lr=0.001) # Optimizer = Adam
```

```python
dataloader = False
for epoch in range(0, 100):
    for x, y in dataloader:
        out = model(x) # Forward Propagation
        loss = loss_fn(out, y) # Loss function
        loss.backward() # Back Propagation
        optimizer.step() # Update parameter based on Gradient Loss
        optimizer.zero_grad() # Reset Gradient
```

# Auto Encoder

```python
class AutoEncoder(nn.Module):
    def __init__(self):
        super().__init__()

        self.encoder = nn.Sequential( # Feature Compression
            nn.Linear(784, 128), # Input Layer 784 to Hidden Layer 128
            nn.ReLU(),
            nn.Linear(128, 32), # Hidden layer 128 to Hidden Layer 32
            nn.ReLU()
        )

        self.decoder = nn.Sequential( # Feature Reconstruction
            nn.Linear(32, 128), # Input Layer 32 to Hidden Layer 128
            nn.ReLU(),
            nn.Linear(128, 784), # Input Layer 128 to Output Layer 784
            nn.Sigmoid()
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)
```

```python
data = False
for epoch in range(100):
    for x, _ in data:
        x = x.view(-1, 784) # Flatten to 784
        out = model(x) # Forward Propagation
        loss = loss_fn(out, x) # Loss function
        loss.backward() # Back propagation
        optimizer.step() # Optimize parameter
        optimizer.zero_grad() # Reset gradient
```
